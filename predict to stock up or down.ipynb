```python
df=pd.read_csv("Market.csv.zip")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# 2. Date column ko datetime mein convert karna aur index set karna
df['Date'] = pd.to_datetime(df['Date'])
df.set_index('Date', inplace=True)

# 3. Target Variable (y) banana: Hum agle din (Next Day) ka price predict karenge
# shift(-1) ka matlab hai 'Close' column ki values ek row upar shift ho jayengi
df['Next_Day_Close'] = df['Close'].shift(-1)

# 4. Missing values drop karna
# shift(-1) ki wajah se dataset ki aakhri row mein NaN aa jayega, isliye usko drop karna zaroori hai
df.dropna(inplace=True)

# 5. Features (X) aur Target (y) define karna
# Hum 'Index' aur 'Adj Close' ko drop kar rahe hain taaki model simple rahe
X = df[['Open', 'High', 'Low', 'Close', 'Volume']]
y = df['Next_Day_Close']

# 6. Train-Test Split
# DHYAN DEIN: Stock data time-series hota hai, isliye shuffle=False rakhna zaroori hai!
# Varna model future data ko train karne mein use kar lega (Data Leakage).
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

# 7. Model initialize aur train karna
model = LinearRegression()
model.fit(X_train, y_train)

# 8. Test data par predictions karna
predictions = model.predict(X_test)

# 9. Model ki accuracy aur error check karna
accuracy = model.score(X_test, y_test)
print(f"Model Accuracy (R^2 Score): {accuracy * 100:.2f}%")
```

```python
import joblib
joblib.dump(model, 'stock_model.pkl')
```
